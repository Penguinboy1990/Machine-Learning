  I used a median imputation strategy for the numeric variables and mode imputation strategy for categorical variables. I used median because medical data is often skewed by clinical outliers. This makes the median more accurate than the mean since the mean would be affected by the skew. For the categorical variables, I used mode or most_frequent imputation. The categories cannot be averaged numerically because categorical data doesn't work that way, even when represented by numbers. This preserves the dominant clinical classification category without introducing invalid values.

4. Discuss how tuning hyperparameters such as regularization parameter C and kernel coefficient gamma in an SVC impacts the trade-off between bias, variance, and model generalization on unseen medical test data.

  Tuning hyperparameters balances between bias and variance by controlling how strongly the model fits. High C penalizes misclassifications whlie high gamma restricts each support vector's radius of influence, which lowers bias but increases variance. This leads to jagged decision boundaries that can overfit on the training data then fail on new external patient data because it follows the training data too closely to accurately predict unseen data. Low gamma values will increase bias and underfitting, which does not generalize well.
