3) Explain how the Scikit-Learn ColumnTransformer and Pipeline streamline the data preparation process and prevent data leakage across preprocessing steps.

  Pipelines let us chain transforms, like imputation and scaling, into one single object. The ColumnTransformer applies different steps to specific feature types at the same time, such as StandardScaler on numeric data and OneHotEncoder for categorical data. This removes some manual steps and also maintains exact matrix alignment. Data leakage occurs when the test data set statistics, like mean or variance, leaks into the training. The Pipelines prevent this by computing the parameters purely on the training set during .fit() and then applying those exact metrics to test sets during the .transform() so the data is isolated.
