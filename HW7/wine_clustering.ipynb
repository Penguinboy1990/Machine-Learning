Unsupervised Chemical Segmentation on the Wine Recognition Dataset

Imports

```python
from sklearn.datasets import load_wine
from sklearn.preprocessing import StandardScaler
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
from sklearn.decomposition import PCA
from sklearn.metrics import pairwise_distances
from sklearn.cluster import DBSCAN
import hdbscan
from scipy.cluster.hierarchy import linkage, dendrogram
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.preprocessing import OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.metrics import accuracy_score, silhouette_score
import warnings
```

Task 1

Print dataset dimensions, feature summary statistics, and missing value checks. Standardize all 13 features and display a comparison of feature distributions before and after scaling.

```python
# Load the wine dataset
wine = load_wine(as_frame=True)
X = wine.data
y = wine.target

# Dataset dimensions
print("Dataset Shape (Samples, Features):", X.shape)
print("Target Classes:", wine.target_names)
print("\nClass Distribution:")
print(pd.Series(y).value_counts().sort_index())

# Summary statistics
print("\nSummary Statistics")
print(X.describe())

# Missing value check
print("\nMissing Values:")
print(X.isnull().sum())

# Standardizes all 13 features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
X_scaled = pd.DataFrame(X_scaled, columns=X.columns, index=X.index)

# Compares distributions before and after scaling
comparison = pd.concat(
    [X.describe().add_suffix('_Original'),
     X_scaled.describe().add_suffix('_Scaled')],
    axis=1)

print("\nFeature Distributions Before and After Scaling:")
print(comparison)
print("\nFirst Five Rows:")
display(X.head())
print("\nFirst Five Scaled Rows:")
display(X_scaled.head())
```

Task 2

Iterate k from 1 to 10 for K-Means clustering on the scaled features. Compute and plot side-by-side:
1) The **Elbow Method** curve.
2) The **Silhouette Scores** curve for k in [2, 10]. **Print the optimal k value suggested by both metrics.**

```python
# Result storage
inertia = []
silhouette_scores = []

# Elbow Method
for k in range(1, 11):
  kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
  kmeans.fit(X_scaled)
  inertia.append(kmeans.inertia_)

# Silhouette Method
for k in range(2, 11):
  kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
  labels = kmeans.fit_predict(X_scaled)
  silhouette_scores.append(silhouette_score(X_scaled, labels))

# Estimates the elbow using maximum distance from the endpoint line
x_vals = np.array(list(range(1, 11)))
y_vals = np.array(inertia)

# Normalizes both axes
x_norm = (x_vals - x_vals.min()) / (x_vals.max() - x_vals.min())
y_norm = (y_vals - y_vals.min()) / (y_vals.max() - y_vals.min())

# Distance from straight line connecting endpoints
distances = np.abs(
    (y_norm[-1] - y_norm[0]) * x_norm
    - (x_norm[-1] - x_norm[0]) * y_norm
    + x_norm[-1] * y_norm[0]
    - y_norm[-1] * x_norm[0]
) / np.sqrt(
    (y_norm[-1] - y_norm[0])**2
    + (x_norm[-1] - x_norm[0])**2)

elbow_k = x_vals[np.argmax(distances)]

# Best Silhouette Score
silhouette_k = np.argmax(silhouette_scores) + 2

# Plots both methods side by side
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Elbow curve
axes[0].plot(x_vals, inertia, marker='o', linestyle='--', color='green')
axes[0].axvline(elbow_k, color='lime', linestyle=':', label=f'Estimated Elbow k={elbow_k}')
axes[0].set_title('The Elbow Method (Inertia)')
axes[0].set_xlabel('Number of Clusters (k)')
axes[0].set_ylabel('Inertia')
axes[0].legend()
axes[0].grid(True)

# Silhouette curve
axes[1].plot(range(2, 11), silhouette_scores, marker='o', linestyle='--', color='red')
axes[1].axvline(silhouette_k, color='red', linestyle=':', label=f'Best Silhouette k={silhouette_k}')
axes[1].set_title('Silhouette Scores for Various k')
axes[1].set_xlabel('Number of Clusters (k)')
axes[1].set_ylabel('Silhouette Score')
axes[1].legend()
axes[1].grid(True)
plt.tight_layout()
plt.show()

print("Optimal k suggested by Elbow Method:", elbow_k)
print("Optimal k suggested by Silhouette Method:", silhouette_k)

if elbow_k == silhouette_k:
    print("Both methods suggest k =", elbow_k)
else:
    print("The two methods suggest different values of k.")
```

Task 3

Apply PCA strictly to reduce the 13 scaled features down to 2 principal components for visualization purposes. Fit a K-Means model (k=3) on the 2D projected space. Plot the decision boundaries using the **Voronoi** method.

```python
# Applies PCA to reduce 13 features to 2 components
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

print("Original Dimensions:", X_scaled.shape)
print("PCA Dimensions:", X_pca.shape)
print("\nExplained Variance Ratio: ", pca.explained_variance_ratio_)
print("\nTotal Explained Variance: ", sum(pca.explained_variance_ratio_))

# Fits K-Means on 2D PCA data
k = 3
kmeans_pca = KMeans(n_clusters=k, random_state=42, n_init=10)
pca_labels = kmeans_pca.fit_predict(X_pca)

# Extracts centroids
centroids_pca = kmeans_pca.cluster_centers_

# Creates mesh grid
x_min, x_max = X_pca[:, 0].min() - 1, X_pca[:, 0].max() + 1
y_min, y_max = X_pca[:, 1].min() - 1, X_pca[:, 1].max() + 1
xx, yy = np.meshgrid(np.linspace(x_min, x_max, 300), np.linspace(y_min, y_max, 300))

# Predicts cluster labels for mesh points
grid_points = np.c_[xx.ravel(), yy.ravel()]
Z = kmeans_pca.predict(grid_points)
Z = Z.reshape(xx.shape)

# Plots decision boundaries
plt.figure(figsize=(10, 7))
plt.contourf(xx, yy, Z, alpha=0.3, cmap='tab10')

# Plots PCA data points
sns.scatterplot(x=X_pca[:, 0], y=X_pca[:, 1], hue=pca_labels, palette='tab10', s=70, alpha=0.8, edgecolor='k')

# Plots centroids
plt.scatter(centroids_pca[:, 0], centroids_pca[:, 1], marker='X', s=250, linewidths=2, color='red', label='Centroids')
plt.title('PCA + K-Means Voronoi Decision Boundaries (k=3)')
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.grid(True)
plt.tight_layout()
plt.show()
```

Task 4

Select 5 representative wine samples from the dataset. Using the fitted 13-feature K-Means model (k=3), compute the distance matrix. Generate and display a heatmap visualizing the soft clustering distance profiles across all 3 cluster centroids, highlighting the minimum distance corresponding to each sample's hard cluster assignment.

```python
# Fits K-Means on all 13 standard features
kmeans_13 = KMeans(n_clusters=k, random_state=42, n_init=10)
kmeans_labels = kmeans_13.fit_predict(X_scaled)

# Select five representative samples
sample_indices = [0, 35, 70, 105, 177]
representative_samples = X_scaled.iloc[sample_indices]

# Calculates the distances to all three centroids
distance_matrix = pairwise_distances(representative_samples, kmeans_13.cluster_centers_, metric='euclidean')

# Distance DataFrame
distance_df = pd.DataFrame(distance_matrix,
    index=[f'Wine Sample {i}' for i in sample_indices],
    columns=['Cluster 0', 'Cluster 1', 'Cluster 2'])
print("Distance Matrix (5 Samples x 3 Centroids):")
print(distance_df.round(3))

# Identifies the nearest centroid for each sample
nearest_clusters = np.argmin(distance_matrix, axis=1)
print("\nHard Cluster Assignments:")
for i, cluster in enumerate(nearest_clusters):
    print(f"Wine Sample {sample_indices[i]}: Cluster {cluster}")

# Plots the heatmap
plt.figure(figsize=(10, 5))
ax = sns.heatmap(distance_df, annot=True, fmt='.2f', cmap='Spectral', linewidths=1, cbar_kws={'label': 'Euclidean Distance'})

# Highlights minimum distance in each row
for row in range(len(distance_matrix)):
    col = nearest_clusters[row]
    ax.add_patch(plt.Rectangle((col, row), 1, 1, fill=False, edgecolor='red', linewidth=3))
plt.title('K-Means Distance Profiles: Five Wine Samples')
plt.xlabel('Cluster Centroids')
plt.ylabel('Wine Samples')
plt.tight_layout()
plt.show()

```

Task 5

Fit a *DBSCAN* model on the 13-dimensional scaled feature space. Experiment with the two needed hyperparameter settings. Print the number of identified clusters and noise points. Visualize the DBSCAN assignments on the 2D PCA projection space, clearly distinguishing noise points from core cluster members.

```python
# Two DBSCAN configurations
dbscan_configs = [
    {'eps': 1.6, 'min_samples': 5},
    {'eps': 2.1, 'min_samples': 5},
    {'eps': 2.4, 'min_samples': 5},
    {'eps': 2.9, 'min_samples': 5},
    {'eps': 1.6, 'min_samples': 6},
    {'eps': 2.1, 'min_samples': 6},
    {'eps': 2.4, 'min_samples': 6},
    {'eps': 2.9, 'min_samples': 6}]
dbscan_results = {}
for i, params in enumerate(dbscan_configs, start=1):
    dbscan = DBSCAN(eps=params['eps'], min_samples=params['min_samples'])
    labels = dbscan.fit_predict(X_scaled)
    dbscan_results[i] = labels
    n_clusters = len(set(labels)) - (1 if -1 in labels else 0)
    n_noise = list(labels).count(-1)
    print(f"\nDBSCAN Configuration {i}")
    print("Parameters:", params)
    print("Number of Clusters:", n_clusters)
    print("Number of Noise Points:", n_noise)

# Selects one of the configurations for visualization purposes
dbscan_labels = dbscan_results[2]

# Visualizes on PCA projection
plt.figure(figsize=(10, 6))

# Plots each cluster and noise separately
unique_labels = sorted(set(dbscan_labels))
for label in unique_labels:
    mask = dbscan_labels == label
    if label == -1:
        plt.scatter(X_pca[mask, 0], X_pca[mask, 1], c='red', marker='x', s=80, label='Noise')
    else:
        plt.scatter(X_pca[mask, 0], X_pca[mask, 1], s=65, alpha=0.8, label=f'Cluster {label}')
plt.title('DBSCAN Clustering on PCA Projection')
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

Task 6

Fit an *HDBSCAN* model (using *min_cluster_size=5* or *min_samples=5*) on the 13-dimensional scaled feature space. Print the total number of clusters discovered and noise points detected. Visualize the resulting HDBSCAN cluster assignments on the 2D PCA projection space.

```python
# Fits HDBSCAN on the 13-dimensional standardized data
hdbscan_model = hdbscan.HDBSCAN(min_cluster_size=5, min_samples=5)
hdbscan_labels = hdbscan_model.fit_predict(X_scaled)

# Counts clusters and noise
n_clusters_hdbscan = len(set(hdbscan_labels) - {-1})
n_noise_hdbscan = list(hdbscan_labels).count(-1)
print("HDBSCAN Results")
print("Number of Clusters:", n_clusters_hdbscan)
print("Number of Noise Points:", n_noise_hdbscan)
plt.figure(figsize=(10, 6))
unique_labels = sorted(set(hdbscan_labels))
for label in unique_labels:
    mask = hdbscan_labels == label
    if label == -1:
        plt.scatter(X_pca[mask, 0], X_pca[mask, 1], c='lime', marker='x', s=80, label='Noise')
    else:
        plt.scatter(X_pca[mask, 0], X_pca[mask, 1], s=65, alpha=0.8, label=f'Cluster {label}')
plt.title('HDBSCAN Clustering on PCA Projection')
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

Task 7

Compute a single-linkage hierarchical clustering matrix on the scaled Wine features. Plot a hierarchical clustering **dendrogram** to visualize how individual chemical profiles merge across distance thresholds, connecting the visual structure to HDBSCAN's stability extraction mechanics.

```python
# Computes the single-linkage hierarchical clustering
linked = linkage(X_scaled, method='single', metric='euclidean')

# Plots the dendrogram
plt.figure(figsize=(20, 8))
dendrogram(linked, orientation='top', distance_sort='descending', leaf_rotation=90,
           leaf_font_size=7, color_threshold=0.7 * max(linked[:, 2]))
plt.title('Hierarchical Clustering Dendrogram (Single Linkage Tree)')
plt.xlabel('Wine Sample Index')
plt.ylabel('Euclidean Distance')
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()
```

Task 8

Treat the cluster assignments generated by **K-Means (k=3), DBSCAN, and HDBSCAN** as categorical feature representations or pseudo-labels. Perform a stratified train/test split (80/20) on the Wine dataset and conduct an experiment: train a supervised classifier such as a **Random Forest or Logistic Regression** using these cluster labels to predict the actual ground-truth wine cultivars (Class 0, 1, 2). ***Plot the classification accuracy and Silhouette Scores for all three algorithms.***

```python
cluster_assignments = {'K-Means': kmeans_labels, 'DBSCAN': dbscan_labels, 'HDBSCAN': hdbscan_labels}
accuracy_results = {}
silhouette_results = {}
indices = np.arange(len(y))
train_idx, test_idx = train_test_split(indices, test_size=0.2, random_state=35, stratify=y)

for algorithm, labels in cluster_assignments.items():
    print(f"{algorithm} Results:")

    cluster_features = pd.DataFrame({'Cluster': labels.astype(str)})
    X_train = cluster_features.iloc[train_idx]
    X_test = cluster_features.iloc[test_idx]
    y_train = y.iloc[train_idx]
    y_test = y.iloc[test_idx]

    classifier = Pipeline([
        ('encoder', OneHotEncoder(handle_unknown='ignore')),
        ('model', RandomForestClassifier(n_estimators=200, random_state=42, class_weight='balanced'))])
    classifier.fit(X_train, y_train)

    y_pred = classifier.predict(X_test)
    accuracy = accuracy_score(y_test, y_pred)
    accuracy_results[algorithm] = accuracy
    valid_mask = labels != -1
    valid_labels = labels[valid_mask]
    valid_features = X_scaled[valid_mask]
    n_clusters = len(np.unique(valid_labels))

    if 2 <= n_clusters < len(valid_labels):
        sil_score = silhouette_score(valid_features, valid_labels)
    else:
        sil_score = np.nan

    silhouette_results[algorithm] = sil_score
    print(f"Accuracy: {accuracy:.2f}")
    print(f"Silhouette Score: {sil_score:.2f}")
    print()

# Creates the results DataFrame
results_df = pd.DataFrame({
    'Algorithm': list(cluster_assignments.keys()),
    'Classification Accuracy': [accuracy_results[a] for a in cluster_assignments],
    'Silhouette Score': [silhouette_results[a] for a in cluster_assignments]})
print("\nFinal Evaluation Results:")
print(results_df)

# Plots classification accuracy and silhouette scores:
# Creates the side-by-side comparison plots
fig, axes = plt.subplots(1, 2, figsize=(13, 5))

# Classification accuracy
sns.barplot(data=results_df, x='Algorithm', y='Classification Accuracy',
            hue='Algorithm', palette='viridis', legend=False, ax=axes[0])

axes[0].set_title('Classification Accuracy Comparison')
axes[0].set_ylabel('Accuracy')
axes[0].set_xlabel('Clustering Algorithm')
axes[0].set_ylim(0, 1)

for container in axes[0].containers:
    axes[0].bar_label(container, fmt='%.3f')

# Silhouette score
sns.barplot(data=results_df, x='Algorithm', y='Silhouette Score',
            hue='Algorithm', palette='magma', legend=False, ax=axes[1])
axes[1].set_title('Silhouette Score Comparison')
axes[1].set_ylabel('Silhouette Score')
axes[1].set_xlabel('Clustering Algorithm')

for container in axes[1].containers:
    axes[1].bar_label(container, fmt='%.3f')
plt.tight_layout()
plt.show()
```
