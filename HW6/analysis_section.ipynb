3) Why is it critical to fit the PCA transformer strictly on the training split rather than on the entire dataset prior to splitting? What negative impacts occur if PCA is fitted on the full dataset before training downstream classifiers?

The PCA transformer must be fitted strictly on the training data rather than the entire set to prevent data leakage. Fitting on the whole set would allow test images to influence the calculated mean and principal components, even  when their labels are not used. This would make the evaluation less representative of real world performance since information from the test set would influence preprocessing. The proper approach is to fit the PCA on the training data and use the same transformation on the test data so that we can gague how well the model reacts to new or unfamiliar inputs.

5) Based on your findings in Task 8 regarding accuracy and training efficiency, why are global linear transformations like PCA and classical pixel classifiers insufficient for complex computer vision tasks?

PCA and Classical Classifiers perform well on controlled images. But PCA is a global linear transformer. This means it does not do well at learning spatial relationships between pixels. Classical classifiers depend on the quality of their inputs and can struggle with certain changes in the image, such as the pose, lighting, or expression. Using a more complex model does not mean that these hierarchical spatial features are automatically learned. Convolutional Neural Networks (CMMs) fix these limitations by learning local patterns, like edges and textures. They combine them into higher-level visual features. This makes them a better fit for computer vision tasks. But, they also need more data and computational resources. PCA gives us great dimensionality reduction, but deep-learning architectures are much better for complex images. They are capable of learning spatially structured features directly from the data.
